```python
df_clean=df[(df['Customer ID'].notnull()) &
    (df['Quantity']>0) &
    (df['StockCode']!='B')]
```

```python
df_clean=df_clean[df_clean['Price']>0]
```

```python
df_clean['Total_Price']=df_clean['Quantity'] * df_clean['Price']
```

```python
import datetime as dt

reference_date=df_clean['InvoiceDate'].max() + dt.timedelta(days=1)  #it gives date after latest purchase

rfm = df_clean.groupby('Customer ID').agg({
    'InvoiceDate' : lambda x:(reference_date -x.max()).days	,
    'Invoice' : 'nunique' ,
    'Total_Price' :'sum'})
rfm.columns=['Recency' ,'Frequency' ,'Monetory']
rfm.head()
```

```python
rfm.describe()
```

```python
import numpy as np

rfm_log=rfm.copy()
rfm_log['Monetory']= np.log1p(rfm_log['Monetory'])
rfm_log['Frequency']= np.log1p(rfm_log['Frequency'])
```

```python
rfm[['Recency','Frequency','Monetory']].hist(figsize=(10,6))
rfm_log[['Recency','Frequency','Monetory']].hist(figsize=(10,6))
```

```python
from sklearn.preprocessing import StandardScaler

scaler= StandardScaler()
rfm_scaled=scaler.fit_transform(rfm_log)

rfm_scaled=pd.DataFrame(rfm_scaled, index=rfm_log.index ,columns=['Recency' ,' Frequency' ,'Monetory'])
rfm_scaled.head()
```

```python
rfm_scaled.describe()
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

wcss=[]
K_range=range(2,11)

for k in K_range:
    kmeans=KMeans(n_clusters=k, random_state=42 ,n_init=10)
    kmeans.fit(rfm_scaled)
    wcss.append(kmeans.inertia_)

plt.plot(K_range ,wcss ,marker='o')
plt.xlabel('Number of clusters(k)')
plt.ylabel('WCSS')
plt.title('Elbow Method')
plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores=[]

for k in K_range:
    kmeans=KMeans(n_clusters=k , random_state=42, n_init=10)
    labels=kmeans.fit_predict(rfm_scaled)
    score=silhouette_score(rfm_scaled,labels)
    silhouette_scores.append(score)

plt.plot(K_range,silhouette_scores,marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score Method')
plt.show()

                
```

```python
kmeans=KMeans(n_clusters=5, random_state=42,n_init=10)
rfm['Cluster']=kmeans.fit_predict(rfm_scaled)
rfm.head()
```

```python
rfm.groupby('Cluster')[['Recency','Frequency','Monetory']].mean()
```

```python
from sklearn.decomposition import PCA
import seaborn as sns

pca=PCA(n_components=2)
pca_result=pca.fit_transform(rfm_scaled)

rfm['PCA1']=pca_result[:,0]
rfm['PCA2']=pca_result[:,1]

plt.figure(figsize=(8,6))
sns.scatterplot(data=rfm ,x='PCA1' , y='PCA2' ,hue='Cluster' ,palette='tab10')
plt.title('Customer Segments(PCA Visualization)')
plt.show()
```
